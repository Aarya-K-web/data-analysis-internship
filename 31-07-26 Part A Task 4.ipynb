```python
def calculate_average(numbers):
    total = 0

    for num in numbers:
        total += num

    average = total / len(numbers)
    return average


nums = [10, 20, 30, 40, 50]

avg = calculate_average(nums)

print("Average is:", avg)

if avg > 25:
    print("Average is greater than 25")
else:
    print("Average is 25 or less")
```

```python
# Original code: uses an explicit loop to accumulate the total.
# Optimized code: sum() is clearer and implemented efficiently.
# Both versions produce the same output.

def calculate_average(numbers):
    return sum(numbers) / len(numbers)


nums = [10, 20, 30, 40, 50]
avg = calculate_average(nums)

print("Average is:", avg)

if avg > 25:
    print("Average is greater than 25")
else:
    print("Average is 25 or less")
```
