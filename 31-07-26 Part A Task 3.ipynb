# Mistakes found

1. `def calculate_average(numbers)` needs a colon (`:`) at the end.
2. `for num in numbers` needs a colon at the end.
3. `len(number)` refers to an undefined variable; it should be `len(numbers)`.
4. `"Average is: " + avg` tries to join a string and a float. Use an f-string or convert `avg` to a string.
5. `if avg > 25` needs a colon at the end.
6. `else` needs a colon at the end.
7. `else` must be aligned with its matching `if`.


```python
# Program to calculate the average of numbers

def calculate_average(numbers):
    total = 0

    for num in numbers:
        total += num

    average = total / len(numbers)
    return average


nums = [10, 20, 30, 40, 50]

avg = calculate_average(nums)

print(f"Average is: {avg}")

if avg > 25:
    print("Average is greater than 25")
else:
    print("Average is 25 or less")
```
