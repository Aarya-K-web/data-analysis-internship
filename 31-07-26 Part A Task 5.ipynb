```python
# Standalone Bank Interest & Fee Calculation Function
def calculate_accrued_interest(balance, rate, years, transaction_fee=0.0):
    """
    Calculates compound interest accrued over a specified term after deducting a transaction fee.
    
    Args:
        balance (float/int): Initial account balance (must be >= 0).
        rate (float/int): Annual interest rate as a percentage (e.g., 5 for 5%).
        years (int): Investment duration in years (must be >= 1).
        transaction_fee (float/int, optional): Flat fee deducted from balance before compounding.
        
    Returns:
        float: Total final balance rounded to 2 decimal places.
        
    Raises:
        ValueError: If balance, rate, or years are negative, or if fee exceeds balance.
        TypeError: If non-numeric values are passed.
    """
    if not all(isinstance(val, (int, float)) for val in [balance, rate, years, transaction_fee]):
        raise TypeError("All numeric inputs (balance, rate, years, fee) must be integers or floats.")
        
    if balance < 0:
        raise ValueError("Initial balance cannot be negative.")
    if rate < 0:
        raise ValueError("Interest rate cannot be negative.")
    if years < 1:
        raise ValueError("Investment duration must be at least 1 year.")
    if transaction_fee < 0:
        raise ValueError("Transaction fee cannot be negative.")
    if transaction_fee > balance:
        raise ValueError("Transaction fee cannot exceed the initial balance.")
        
    # Deduct transaction fee
    effective_balance = balance - transaction_fee
    
    # Calculate Compound Interest: A = P * (1 + r/100)^t
    final_balance = effective_balance * ((1 + (rate / 100)) ** years)
    
    return round(final_balance, 2)
```
