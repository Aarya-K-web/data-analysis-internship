```python
# NumPy Array Arithmetic
import numpy as np

array1 = np.array([10, 20, 30, 40])
array2 = np.array([3, 4, 5, 6])

print("Array 1:", array1)
print("Array 2:", array2)
print("Addition:", array1 + array2)
print("Subtraction:", array1 - array2)
print("Multiplication:", array1 * array2)
print("Division:", array1 / array2)
print("Modulus:", array1 % array2)
print("Exponentiation:", array1 ** array2)
print("Floor division:", array1 // array2)

# Broadcasting: Both arrays have shape (4,), so these operations run element by element.
# Broadcasting is not needed here. NumPy can also broadcast compatible shapes,
# such as adding a scalar to every element: array1 + 5.
```
