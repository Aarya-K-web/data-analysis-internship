# NumPy Array Indexing and Slicing

This notebook creates a 5×5 NumPy array and demonstrates common ways to access its values. NumPy uses zero-based indexing: the first row and column have index `0`.

```python
import numpy as np

# Create a 5×5 array containing the values 1 to 25.
array_5x5 = np.arange(1, 26).reshape(5, 5)
array_5x5
```

## Positive indexing
`array_5x5[1, 2]` selects the item at row index `1` and column index `2` (the second row, third column).

```python
positive_index = array_5x5[1, 2]
positive_index
```

## Negative indexing
Negative indices count backward from the end. `array_5x5[-1, -2]` selects the last row and second-last column.

```python
negative_index = array_5x5[-1, -2]
negative_index
```

## Slicing
`array_5x5[1:4, 1:4]` takes rows with indices `1` through `3` and columns with indices `1` through `3`. The stop index (`4`) is excluded.

```python
sliced_array = array_5x5[1:4, 1:4]
sliced_array
```

## Row selection
`array_5x5[2, :]` selects every column (`:`) from row index `2`, which is the third row.

```python
third_row = array_5x5[2, :]
third_row
```

## Column selection
`array_5x5[:, 3]` selects every row (`:`) from column index `3`, which is the fourth column.

```python
fourth_column = array_5x5[:, 3]
fourth_column
```

## Extracting a submatrix
`array_5x5[1:4, 2:5]` extracts a 3×3 submatrix using rows `1`–`3` and columns `2`–`4`.

```python
submatrix = array_5x5[1:4, 2:5]
submatrix
```
