````python
# Student Grade Classifier and Statistics
def calculate_grade_summary(scores):
    passing_grade = 60
    summary = {"passed": 0, "failed": 0, "average": 0.0}
    total_score = 0
    
    for score in scores:
        total_score += score
        if score >= passing_grade:
            summary["passed"] += 1
        else:
            summary["failed"] += 1
            
    summary["average"] = round(total_score / len(scores), 2)
    return summary

# Data Input
student_scores = [85, 42, 76, 95, 58, 63, 90]

# Function Execution
results = calculate_grade_summary(student_scores)

# Display Output
print("--- Student Grade Analysis ---")
print(f"Total Students: {len(student_scores)}")
print(f"Average Score:  {results['average']}")
print(f"Passed:         {results['passed']}")
print(f"Failed:         {results['failed']}")
````

# Explanation: Student Grade Classifier and Statistics

This program analyzes student exam scores. It counts who passed and failed, then calculates the average score.

## Line-by-line explanation

- `# Student Grade Classifier and Statistics` is a comment for people reading the code. Python ignores comments.
- `def calculate_grade_summary(scores):` creates a function named `calculate_grade_summary`. It accepts a list of scores, for example `[85, 42, 76]`.
- `passing_grade = 60` sets the passing mark to 60.
- `summary = {"passed": 0, "failed": 0, "average": 0.0}` creates a dictionary to store the results. Initially, the counts and average are zero.
- `total_score = 0` creates a variable that will hold the sum of all scores.
- `for score in scores:` loops through every score in the list. For `[85, 42, 76]`, it processes 85, then 42, then 76.
- `total_score += score` adds the current score to the total. Example: 0 + 85 = 85, then 85 + 42 = 127.
- `if score >= passing_grade:` checks whether the score is 60 or higher.
- `summary["passed"] += 1` adds one to the passed count when the condition is true.
- `else:` handles scores below 60.
- `summary["failed"] += 1` adds one to the failed count.
- `summary["average"] = round(total_score / len(scores), 2)` calculates the average: total score divided by the number of students, rounded to two decimal places.
- `return summary` sends the completed dictionary back from the function.
- `student_scores = [85, 42, 76, 95, 58, 63, 90]` stores the marks for seven students.
- `results = calculate_grade_summary(student_scores)` runs the function and saves the result in `results`.
- Each `print(...)` line displays a heading or a value. The `f` in `f"..."` lets Python insert values inside curly brackets, such as `{results['average']}`.
