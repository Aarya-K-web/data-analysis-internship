# NumPy Statistical and Position Functions

This notebook demonstrates common NumPy functions for summarizing data and finding the positions of extreme values.

```python
import numpy as np

data = np.array([12, 7, 9, 15, 10, 7, 20])
print('Data:', data)
```

## Calculations

- sum() adds all values.
- mean() returns the arithmetic average.
- median() returns the middle value after sorting.
- std() measures how spread out the values are.
- var() is the variance, equal to the standard deviation squared.
- min() and max() return the smallest and largest values.
- argmin() and argmax() return the zero-based indices of the first smallest and largest values.

```python
print('Sum:', np.sum(data))
print('Mean:', np.mean(data))
print('Median:', np.median(data))
print('Standard deviation:', np.std(data))
print('Variance:', np.var(data))
print('Minimum:', np.min(data))
print('Maximum:', np.max(data))
print('Index of minimum:', np.argmin(data))
print('Index of maximum:', np.argmax(data))
```

For this array, the minimum value (7) first appears at index 1, and the maximum value (20) appears at index 6. NumPy uses population formulas for std() and var() by default (ddof=0).
